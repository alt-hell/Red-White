```python
# ==========================================
# Welcome to the Pattern Generator and Number Analyzer
# ==========================================

print("=" * 50)
print("   Welcome to the Pattern Generator and Number Analyzer")
print("=" * 50)

user_input = int(input(
    "\nSelect an Option:\n"
    "1. Generate a Pattern\n"
    "2. Analyze a Range of Numbers\n"
    "3. Exit\n\n"
    "Enter Your Choice: "
))

# ==========================================
# Pattern Generator
# ==========================================

if user_input == 1:

    pattern = int(input(
        "\nChoose a Pattern:\n"
        "1. Right Angled Triangle\n"
        "2. Pyramid\n"
        "3. Left Angled Triangle\n\n"
        "Enter Your Choice: "
    ))

    rows_pattern = int(input("Enter the number of rows: "))

    # Right Angled Triangle
    if pattern == 1:
        print("\nRight Angled Triangle:\n")

        for i in range(1, rows_pattern + 1):
            for j in range(1, i + 1):
                print("*", end=" ")
            print()

    # Pyramid
    elif pattern == 2:
        print("\nPyramid:\n")

        for i in range(1, rows_pattern + 1):
            for j in range(rows_pattern - i):
                print(" ", end="")
            for k in range(1, 2 * i):
                print("*", end="")
            print()

    # Left Angled Triangle
    elif pattern == 3:
        print("\nLeft Angled Triangle:\n")

        for i in range(1, rows_pattern + 1):
            for j in range(rows_pattern - i):
                print("  ", end="")
            for k in range(1, i + 1):
                print("* ", end="")
            print()

    else:
        print("\nInvalid Pattern Choice!")

# ==========================================
# Number Analyzer
# ==========================================

elif user_input == 2:

    start_no = int(input("Enter the start of the range: "))
    end_no = int(input("Enter the end of the range: "))

    print("\nNumber Analysis:\n")

    for i in range(start_no, end_no + 1):
        if i % 2 == 0:
            print(f"{i} is Even")
        else:
            print(f"{i} is Odd")

# ==========================================
# Exit
# ==========================================

elif user_input == 3:
    print("\nThank you for using the Pattern Generator and Number Analyzer!")

# ==========================================
# Invalid Choice
# ==========================================

else:
    print("\nInvalid Choice! Please run the program again.")
```
