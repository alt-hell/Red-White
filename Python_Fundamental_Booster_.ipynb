```python
print("=" * 50)
print("      Welcome to the User Information Program")
print("=" * 50)

print("\nHello! 👋")
print("This program will collect some basic information about you.")
print("It will then:")
print("1. Store your information using appropriate data types.")
print("2. Display the data type and memory address of each variable.")
print("3. Show a summary of all the collected information.")
print("\nPlease enter the requested information when prompted.\n")
```

```python
#Full Programm Flow:

name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height (in cm): "))
favorite_number = int(input("Enter your favorite number: "))



print("\n----- Variable Information -----")

print(f"Name Data Type: {type(name)}")
print(f"Name Memory Address: {id(name)}\n")

print(f"Age Data Type: {type(age)}")
print(f"Age Memory Address: {id(age)}\n")

print(f"Height Data Type: {type(height)}")
print(f"Height Memory Address: {id(height)}\n")

print(f"Favorite Number Data Type: {type(favorite_number)}")
print(f"Favorite Number Memory Address: {id(favorite_number)}")


print("\n========== Variable Summary ==========")

print(f"Variable Name : name")
print(f"Value         : {name}")
print(f"Data Type     : {type(name)}")
print(f"Memory Address: {id(name)}")

print("--------------------------------------")

print(f"Variable Name : age")
print(f"Value         : {age}")
print(f"Data Type     : {type(age)}")
print(f"Memory Address: {id(age)}")

print("--------------------------------------")

print(f"Variable Name : height")
print(f"Value         : {height}")
print(f"Data Type     : {type(height)}")
print(f"Memory Address: {id(height)}")

print("--------------------------------------")

print(f"Variable Name : favorite_number")
print(f"Value         : {favorite_number}")
print(f"Data Type     : {type(favorite_number)}")
print(f"Memory Address: {id(favorite_number)}")

print("======================================")
```

```python
#Requirement 3
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height in cm: "))

current_year = 2026
birth_year = current_year - age

height_integer = int(height)
height_round = round(height)

print("\n----- User Information -----")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Height: {height}")

print("\n----- Type Conversion -----")
print(f"Original height (float): {height}")
print(f"Height converted to integer: {height_integer}")
print(f'Rounded value of Height : {height_round}')

print("\n----- Calculation -----")
print(f"Estimated Birth Year: {birth_year}")
```
